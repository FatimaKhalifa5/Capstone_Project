### Target Variable

The target column is **`arr_delay`**, which represents the **arrival delay in minutes**.

- A value of **0** means the flight arrived on time.

```python
#Is your target categorical or continuous? Plot the distribution.
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(df3['arr_delay'], bins=50, kde=True)

plt.title("Distribution of Arrival Delay")
plt.xlabel("Arrival Delay (minutes)")
plt.ylabel("Count")

plt.show()
```

### Observation

The target variable **`arr_delay`** is continuous, making this a **regression** problem.

The distribution of arrival delays is **highly right-skewed**. Most flights have arrival delays close to zero, meaning they arrive on time or with only a small delay. However, a small number of flights experience very large delays, resulting in a long right tail and the presence of outliers.

This indicates that the target variable is not normally distributed.
